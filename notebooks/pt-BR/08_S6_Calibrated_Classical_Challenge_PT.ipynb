# S6: desafio clássico calibrado

Este notebook publica o relatório cacheado do S6. O modo padrão é desabilitado. A calibração usa uma divisão interna de `train` e a avaliação externa ocorre uma única vez em `validation`; `test`, `stress` e `monitor` permanecem selados.

```python
from pathlib import Path
import json
import sys

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'dataset' / 'processed' / 'complaints.parquet').exists():
    parent = PROJECT_ROOT.parent
    if parent == PROJECT_ROOT:
        raise FileNotFoundError('Could not find project root')
    PROJECT_ROOT = parent
SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from consumer_complaint_intelligence.s6 import run_s6, run_s6_smoke
from consumer_complaint_intelligence.s6_reporting import load_s6_report_tables

RUN_MODE = 'disabled'
if RUN_MODE not in {'disabled', 'smoke', 'full'}:
    raise ValueError('RUN_MODE must be disabled, smoke, or full')

scientific_cache = PROJECT_ROOT / 'temp' / 's3' / 'scientific.parquet'
artifact = PROJECT_ROOT / 'temp' / 's6' / 's6_results.json'
config = PROJECT_ROOT / 'config' / 's6_calibrated_classical.json'

if RUN_MODE == 'full':
    run_s6(scientific_cache, artifact, config)
elif RUN_MODE == 'smoke':
    run_s6_smoke(scientific_cache, artifact, config)

if artifact.exists():
    cached = json.loads(artifact.read_text(encoding='utf-8'))
    if cached.get('complete') is True:
        evidence = load_s6_report_tables(artifact)
        display(evidence.calibration_summary)
        display(evidence.outer_summary)
        display(evidence.critical_confusions)
        display(evidence.per_class)
        display(evidence.statuses)
        print('S6 report loaded from cache.')
```
