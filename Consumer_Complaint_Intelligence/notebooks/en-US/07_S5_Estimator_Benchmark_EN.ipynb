# S5: estimator benchmark

This notebook publishes evidence from the development benchmark. The default mode is disabled. S5 fits only on 'train' and evaluates only on 'validation'; 'test', 'stress', and 'monitor' remain sealed.

```python
from pathlib import Path
import json
import sys

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'dataset' / 'processed' / 'complaints.parquet').exists():
    parent = PROJECT_ROOT.parent
    if parent == PROJECT_ROOT:
        raise FileNotFoundError('Could not find project root')
    PROJECT_ROOT = parent
SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from consumer_complaint_intelligence.s5 import run_s5, run_s5_smoke
from consumer_complaint_intelligence.s5_reporting import load_s5_report_tables

RUN_MODE = 'disabled'
if RUN_MODE not in {'disabled', 'smoke', 'full'}:
    raise ValueError('RUN_MODE must be disabled, smoke, or full')

scientific_cache = PROJECT_ROOT / 'temp' / 's3' / 'scientific.parquet'
artifact = PROJECT_ROOT / 'temp' / 's5' / 's5_results.json'
config = PROJECT_ROOT / 'config' / 's5_estimator_benchmark.json'
s4_reference = PROJECT_ROOT / 'temp' / 's4' / 's4_results.json'

if RUN_MODE == 'full':
    run_s5(scientific_cache, artifact, config, reference_artifact_path=s4_reference)
elif RUN_MODE == 'smoke':
    run_s5_smoke(scientific_cache, artifact, config)

if artifact.exists():
    cached = json.loads(artifact.read_text(encoding='utf-8'))
    if cached.get('complete') is True:
        evidence = load_s5_report_tables(artifact)
        display(evidence.candidate_summary)
        display(evidence.critical_confusions)
        display(evidence.per_class)
        display(evidence.reference_parity)
        display(evidence.deferred_candidates)
        print('S5 report loaded from cache.')
```
