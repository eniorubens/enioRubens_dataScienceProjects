# S8: Avaliação confirmatória

Este notebook publica apenas a evidência agregada em cache. O modo `full` é uma abertura única do `test`, controlada pelo runner e pelo token de ambiente; `stress` e `monitor` continuam selados.

A visão científica usa um representante mínimo por grupo novo e a visão operacional é diagnóstica, sem participar da decisão.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
SRC_DIR = PROJECT_ROOT / 'src'
sys.path.insert(0, str(SRC_DIR))

from consumer_complaint_intelligence.s8 import run_s8
from consumer_complaint_intelligence.s8_reporting import load_s8_report_tables

RUN_MODE = 'disabled'
assert RUN_MODE in {'disabled', 'smoke', 'full'}
config = PROJECT_ROOT / 'config' / 's8_confirmatory_protocol.json'
result = PROJECT_ROOT / 'temp' / 's8' / 's8_results.json'
manifest = PROJECT_ROOT / 'config' / 's8_results.json'

run_result = run_s8(
    project_root=PROJECT_ROOT,
    config_path=config,
    result_path=result,
    manifest_path=manifest,
    run_mode=RUN_MODE,
)
print(run_result['status'])
if result.exists():
    evidence = load_s8_report_tables(
        result, manifest_path=manifest, config_path=config
    )
    display(evidence.statuses)
    display(evidence.primary_summary)
    display(evidence.per_class)
    display(evidence.confidence_intervals)
    display(evidence.scientific_operational)
    display(evidence.audit_counts)
else:
    print('No S8 full artifact is available; confirmatory access remains deferred.')
```

## Leitura da evidência

Os gates pertencem exclusivamente à tabela `primary_summary`. Os intervalos bootstrap são diagnósticos e a tabela operacional não altera a decisão. O resultado não contém narrativas, IDs, scores individuais ou autorização de deployment.
